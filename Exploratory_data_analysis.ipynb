```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Understand the dataset
df = sns.load_dataset("titanic")
df.to_csv("titanic_original.csv", index=False)
print(df.shape)
df.info()
print(df.describe())
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())

# 2. Clean (rows kept: no passenger ID, so "duplicates" may be different people)
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])
df = df.drop(columns=["deck"])
df.to_csv("titanic_cleaned.csv", index=False)

# 3. Statistics
print("Overall survival:", round(df["survived"].mean(), 3))
for col in ["sex", "pclass", "embarked", "alone"]:
    print(df.groupby(col)["survived"].mean().round(3))
print(df[["survived", "pclass", "age", "fare"]].corr().round(2))

# 4. Visualizations
# Plot 1: survival by sex
df.groupby("sex")["survived"].mean().plot(kind="bar", color=["salmon", "steelblue"])
plt.title("Survival Rate by Sex"); plt.ylabel("Survival rate"); plt.xlabel("Sex")
plt.xticks(rotation=0)
plt.savefig("plot1_survival_sex.png", dpi=150, bbox_inches="tight"); plt.show()

# Plot 2: survival by class
df.groupby("pclass")["survived"].mean().plot(kind="bar", color="seagreen")
plt.title("Survival Rate by Passenger Class"); plt.ylabel("Survival rate"); plt.xlabel("Class")
plt.xticks(rotation=0)
plt.savefig("plot2_survival_class.png", dpi=150, bbox_inches="tight"); plt.show()

# Plot 3: age distribution
plt.hist(df["age"], bins=20, color="steelblue", edgecolor="black")
plt.title("Age Distribution"); plt.xlabel("Age"); plt.ylabel("Passengers")
plt.savefig("plot3_age_hist.png", dpi=150, bbox_inches="tight"); plt.show()

# Plot 4: fare box plot by class
df.boxplot(column="fare", by="pclass")
plt.title("Fare by Passenger Class"); plt.suptitle("")
plt.xlabel("Class"); plt.ylabel("Fare")
plt.savefig("plot4_fare_boxplot.png", dpi=150, bbox_inches="tight"); plt.show()
```
